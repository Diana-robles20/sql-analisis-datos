# **Bed and Breakfast**

## **Context**

A Bed and Breakfast (BnB) is a type of short-term accommodation, similar to a hotel. With the rise of Airbnb, many people can now rent out their properties to tourists.

In this project, we analyze data from Boston using the bnb.db database to understand how Airbnb has influenced the city’s tourism.

## **Database Schema**

The database consists of three main entities:

**Listing:** represents each property available on Airbnb

**Review:** contains user reviews for each listing

**Availability:** stores information about when a listing is available

A Listing can have multiple Reviews and multiple Availability records over time

## **Tables Description**

### **listings**

The `listings` table contains the following columns:

- **id**: unique identifier of the listing  

- **property_type**: type of the listing (e.g., "Entire rental unit", "Private room in rental unit")  

- **host_name**: Airbnb username of the host  

- **accommodates**: maximum number of occupants  

- **bedrooms**: number of bedrooms  

- **description**: listing description on Airbnb  

---

### **reviews** 

The `reviews` table contains the following columns:

- **id**: unique identifier of the review  

- **listing_id**: ID of the listing that received the review  

- **date**: date the review was posted  

- **reviewer_name**: Airbnb username of the reviewer  

- **comments**: content of the review  

---

### **availabilities**

The `availabilities` table contains the following columns:

- **id**: unique identifier of the availability record  

- **listing_id**: ID of the associated listing  

- **date**: date of availability  

- **available**: whether the listing is available (TRUE or FALSE)  

- **price**: price for staying on that date  


```python
import pandas as pd 
import sqlite3
conn = sqlite3.connect("bnb.db")
```

```python
def show_table(name):
    query = f"SELECT*FROM {name} LIMIT 5;"
    return pd.read_sql(query,conn)
```


## **Creating Views for Analysis**

To better understand the data, we will create **SQL views**. 

By using views, we can extract focused subsets of the data, allowing us to isolate specific aspects of interest while filtering out irrelevant information. This makes it easier to work with targeted data, analyze patterns, and generate more precise insights.


### **View: available**

Dates that are available at all listings. The view contains the following columns:

**id**, which is the id of the listing 

**property_type**, from the listings table.

**host_name**, from the listings table.

**date**, from the availabilities table

```python
query = """ 

DROP VIEW IF EXISTS "available";

CREATE VIEW "available" AS
SELECT "listing_id" AS "id", "property_type","host_name","date"
FROM "listings" JOIN "availabilities" ON
"listings"."id" = "availabilities"."listing_id"
WHERE "availabilities"."available" = 'TRUE';

"""
conn.executescript(query)
```

```python
show_table("available")
```

### **View: frequently_reviewed**

This view contains the 100 most frequently reviewed listings, sorted from most to least frequently reviewed. The view contains the following columns:

**id**, which is the id of the listing from the listings table.

**property_type**, from the listings table.

**host_name**, from the listings table.

**reviews**, which is the number of reviews the listing has received.

```python
query = """ 

DROP VIEW IF EXISTS "frequently_reviewed";

CREATE VIEW "frequently_reviewed" AS
SELECT "listings"."id", "property_type", "host_name",
COUNT("reviews") AS "reviews"
FROM "listings" JOIN "reviews" ON 
"listings"."id" = "reviews"."listing_id"
GROUP BY "listings"."id"
ORDER BY "reviews" DESC, "property_type" ASC, "host_name" ASC LIMIT 100;

"""

conn.executescript(query)
```

```python
show_table("frequently_reviewed")
```

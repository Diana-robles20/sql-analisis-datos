# **Players**

This dataset contains information about baseball players, including personal details, physical attributes, and career timelines. 

### **Table: players**

- **`id`:** unique identifier for each player  
- **`first_name`:** player's first name  
- **`last_name`:** player's last name  
- **`birth_city`:** city of birth  
- **`birth_state`:** state of birth  
- **`birth_country`:** country of birth  
- **`bats`:** batting hand (L = left, R = right)  
- **`throws`:** throwing hand (L = left, R = right)  
- **`height`:** height of the player  
- **`weight`:** weight of the player  
- **`debut`:** date of first professional game  
- **`final_game`:** date of last professional game  

```python
import pandas as pd 
import sqlite3
conn = sqlite3.connect("players.db")
```

### **BASIC FILTERS**

Place of birth of Jackie Robinson

```python
query = """ 

SELECT birth_city, birth_state, birth_country
FROM players
WHERE first_name = 'Jackie'
AND last_name = 'Robinson';

"""
pd.read_sql(query,conn)
```

Batting hand of Babe Ruth


```python
query = """ 

SELECT bats
FROM players
WHERE first_name = 'Babe'
AND last_name = 'Ruth';

"""

pd.read_sql(query,conn)
```

Players without debut date (missing data)

```python
query = """ 

SELECT id
FROM players
WHERE debut IS NULL;

"""
pd.read_sql(query,conn)
```

### **FILTERING + SORTING**

Players born outside the USA

```python
query = """ 

SELECT first_name, last_name
FROM players
WHERE birth_country != 'USA'
ORDER BY first_name ASC, last_name ASC;

"""
pd.read_sql(query,conn)
```

Players who bat right-handed


```python
query = """ 

SELECT first_name, last_name
FROM players
WHERE bats = 'R'
ORDER BY first_name ASC, last_name ASC;

"""
pd.read_sql(query,conn)
```

Players born in Pittsburgh, PA ordered by debut


```python
query = """ 

SELECT first_name, last_name, debut
FROM players
WHERE birth_city = 'Pittsburgh'
AND birth_state = 'PA'
ORDER BY debut DESC, first_name ASC, last_name ASC;

"""

pd.read_sql(query,conn)
```

### **DATE FILTERING**

Players who debuted from the year 2000 onwards

```python
query = """ 

SELECT ROUND(AVG(height), 2) AS 'Average height',
       ROUND(AVG(weight), 2) AS 'Average weight'
FROM players
WHERE debut >= '2000%';

"""

pd.read_sql(query,conn)
```

Players whose last game was in 2022

```python
query = """ 

SELECT first_name, last_name
FROM players
WHERE final_game LIKE '2022%'
ORDER BY first_name ASC, last_name ASC;

"""

pd.read_sql(query,conn)
```

### **AGGREGATIONS**

Players who bat and throw with different hands

**Note:**

**bats != throws** would also include cases where one value is NULL

The correct approach is to explicitly define valid combinations: (L,R) and (R,L)

```python
query = """ 

SELECT COUNT(id) AS "players opposite hand"
FROM players
WHERE (bats = 'L' AND throws = 'R')
   OR (bats = 'R' AND throws = 'L');

"""

pd.read_sql(query,conn)
```

### **ADVANCED ANALYSIS**

Players taller than the average

```python
query = """ 

SELECT birth_country, first_name, last_name, height
FROM players
WHERE height > (
       SELECT AVG(height)
       FROM players
       )
ORDER BY height DESC, first_name ASC, last_name ASC
LIMIT 10;

"""

pd.read_sql(query,conn)
```
